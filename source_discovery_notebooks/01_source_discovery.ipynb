```python
import boto3
import pandas as pd

BUCKET = "cct-ds-code-challenge-input-data"
REGION = "af-south-1"

s3 = boto3.client(
    "s3",
    region_name=REGION
)

response = s3.list_objects_v2(Bucket=BUCKET)

files = []

for obj in response.get("Contents", []):
    files.append({
        "key": obj["Key"],
        "size_bytes": obj["Size"],
        "last_modified": obj["LastModified"]
    })

files_df = pd.DataFrame(files)

files_df
```
