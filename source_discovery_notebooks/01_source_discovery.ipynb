```python
import boto3
import pandas as pd

BUCKET = "cct-ds-code-challenge-input-data"
REGION = "af-south-1"

s3 = boto3.client(
    "s3",
    region_name=REGION
)

response = s3.list_objects_v2(Bucket=BUCKET)

files = []

for obj in response.get("Contents", []):
    files.append({
        "key": obj["Key"],
        "size_bytes": obj["Size"],
        "last_modified": obj["LastModified"]
    })

files_df = pd.DataFrame(files)

files_df
```

```python
files_df["file_type"] = (
    files_df["key"]
    .str.lower()
    .str.split(".")
    .str[-1]
)

files_df["file_type"].value_counts()
```

```python
required_sources = [
    "sr.csv.gz",
    "sr_hex.csv.gz",
    "sr_hex_truncated.csv",
    "city-hex-polygons-8.geojson",
    "city-hex-polygons-8-10.geojson",
]

source_check = []

for key in required_sources:
    try:
        response = s3.head_object(
            Bucket=BUCKET,
            Key=key
        )

        source_check.append({
            "key": key,
            "exists": True,
            "size_bytes": response["ContentLength"],
            "last_modified": response["LastModified"],
        })

    except s3.exceptions.ClientError as exc:
        source_check.append({
            "key": key,
            "exists": False,
            "size_bytes": None,
            "last_modified": None,
            "error": exc.response["Error"]["Code"],
        })

source_check_df = pd.DataFrame(source_check)

source_check_df
```

```python
required_sources = [
    "city-hex-polygons-8-10.geojson",
    "city-hex-polygons-8.geojson",
    "sr.csv.gz",
    "sr_hex.csv.gz",
    "sr_hex_truncated.csv",
]

source_inventory = []

for key in required_sources:
    response = s3.head_object(
        Bucket=BUCKET,
        Key=key
    )

    source_inventory.append({
        "key": key,
        "size_bytes": response["ContentLength"],
        "last_modified": response["LastModified"],
    })

source_inventory_df = pd.DataFrame(source_inventory)

source_inventory_df
```
