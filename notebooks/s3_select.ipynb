```python
import boto3
import pandas as pd

BUCKET = "cct-ds-code-challenge-input-data"
REGION = "af-south-1"

s3 = boto3.client(
    "s3",
    region_name=REGION
)
```

```python
response = s3.select_object_content(
    Bucket=BUCKET,
    Key="city-hex-polygons-8-10.geojson",
    ExpressionType="SQL",
    Expression="""
        SELECT *
        FROM S3Object
    """,
    InputSerialization={
        "JSON": {
            "Type": "DOCUMENT"
        }
    },
    OutputSerialization={
        "JSON": {
            "RecordDelimiter": "\n"
        }
    }
)

records = []

for event in response["Payload"]:
    if "Records" in event:
        records.append(event["Records"]["Payload"])

result = b"".join(records).decode("utf-8")

print(result[:2000])
```
